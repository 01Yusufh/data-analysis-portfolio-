# Distribution of Sales

```python
plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=30)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

## Observation

The distribution of sales is highly right-skewed. Most sales transactions are of relatively low value, while only a few transactions have very high sales.

## Business Insight

Most revenue comes from many low-value transactions, while a few high-value orders contribute significantly to total sales.

## Business Recommendation

Identify the products and customers responsible for high-value orders and develop strategies to encourage more large purchases.

# Top 10 Categories by Sales

```python
top_category = df.groupby("Category")["Sales"].sum()

top_category.plot(kind="bar", figsize=(8,5))
plt.title("Sales by Category")
plt.ylabel("Total Sales")
plt.show()
```

## Observation

Technology has the highest total sales, followed by Furniture, while Office Supplies has the lowest total sales.

## Business Insight

Technology is the company's strongest-performing product category.

## Business Recommendation

Continue investing in Technology products and review pricing or promotional strategies for Office Supplies.

# Sales by Region

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.show()
```

# Monthly Sales Trend

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

monthly_sales.plot(figsize=(10,5))
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

## Observation

Sales fluctuate over time but generally show an upward trend from 2014 to 2017, with the highest sales occurring toward the end of 2017.

## Business Insight

The business appears to be growing, although monthly demand varies.

## Business Recommendation

Plan inventory and staffing for peak sales periods and investigate the causes of seasonal fluctuations.

# Profit by Category

```python
profit_category = df.groupby("Category")["Profit"].sum()

profit_category.plot(kind="bar", figsize=(8,5))

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()
```

## Observation

Technology generated the highest total profit, followed closely by Office Supplies. Furniture recorded the lowest profit among the three categories.

## Business Insight

Although all categories are profitable overall, Technology contributes the most to the company's profitability, while Furniture delivers comparatively weaker financial performance.

## Business Recommendation

Continue investing in Technology products while reviewing the pricing, discounting, and cost structure of Furniture products to improve profitability.

---

# Profit by Sub-Category

```python
profit_subcategory = (
    df.groupby("Sub-Category")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))
colors = ['g' if val >= 0 else 'r' for val in profit_subcategory]
profit_subcategory.plot(kind="bar", color=colors)

plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit")

plt.show()
```

## Observation

Copiers generated the highest profit, followed by Phones and Accessories. Tables recorded a significant loss, while Bookcases and Supplies also contributed relatively low profits.

## Business Insight

Profitability varies considerably across sub-categories. High sales do not necessarily translate into high profit, and some products reduce overall profitability.

## Business Recommendation

Maintain strong inventory and marketing support for high-profit sub-categories such as Copiers and Phones. Investigate the causes of losses in Tables and review pricing, supplier costs, or discount policies.

---

# Profit vs Sales

```python
plt.figure(figsize=(8,6))

plt.scatter(df["Sales"], df["Profit"], alpha=0.5)

plt.title("Profit vs Sales")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.axhline(0, color='red', linestyle='--')
plt.show()
```

## Observation

There is a positive relationship between sales and profit, with higher sales generally leading to higher profits. However, several high-sales transactions still resulted in low or negative profits.

## Business Insight

Increasing sales alone does not guarantee higher profitability. Some high-value transactions may be affected by excessive discounts or high operating costs.

## Business Recommendation

Monitor high-sales orders with low profit margins and implement pricing strategies that improve profitability while maintaining sales growth.

---

# Discount vs Profit

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='Discount', y='Profit', alpha=0.5)
plt.title('Discount vs Profit')
plt.axhline(0, color='red', linestyle='--')
plt.show()
```

## Observation

As discount levels increase, profits tend to decline. Several transactions with high discounts resulted in substantial losses.

## Business Insight

Heavy discounting has a negative impact on profitability and can eliminate the financial benefits of increased sales.

## Business Recommendation

Review discount policies and establish limits on high discounts unless they are supported by clear business objectives such as customer acquisition or inventory clearance.

# Sales & Profit by Customer Segment

Continue investing in customer retention and marketing for the Consumer segment while developing strategies to increase sales and profitability within the Corporate and Home Office segments.

# Sales by Ship Mode

```python
ship_sales = df.groupby("Ship Mode")["Sales"].sum()

plt.figure(figsize=(8,5))
ship_sales.plot(kind="bar")

plt.title("Sales by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Total Sales")

plt.show()
```

## Observation

Standard Class generated the highest total sales by a wide margin. Same Day shipping accounted for the lowest sales.

## Business Insight

Most customers prefer Standard Class shipping, suggesting that they prioritize lower shipping costs over faster delivery.

## Business Recommendation

Continue optimizing Standard Class shipping operations while evaluating whether premium shipping services can be promoted to customers who value faster delivery.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

# Average Shipping Delay

```python
avg_shipping = df.groupby("Ship Mode")["Shipping Days"].mean()

plt.figure(figsize=(8,5))
avg_shipping.plot(kind="bar")

plt.title("Average Shipping Days by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Average Days")

plt.show()
```

## Observation

Standard Class had the longest average delivery time, while Same Day shipping had the shortest delivery time, as expected.

## Business Insight

Shipping performance aligns with the service levels offered by each shipping mode, demonstrating operational consistency.

## Business Recommendation

Monitor shipping performance regularly to ensure delivery times continue to meet customer expectations and service-level commitments.

# Sales by Quarter

```python
quarter_sales = df.groupby("Quarter")["Sales"].sum()

plt.figure(figsize=(8,5))
quarter_sales.plot(kind="bar")

plt.title("Sales by Quarter")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")

plt.show()
```

```python
df["Day"] = df["Order Date"].dt.day_name()
```

```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

sales_day = (
    df.groupby("Day")["Sales"]
      .sum()
      .reindex(day_order)
)

plt.figure(figsize=(9,5))
sales_day.plot(kind="bar")

plt.title("Sales by Day of Week")
plt.xlabel("Day")
plt.ylabel("Total Sales")

plt.show()
```

## Observation

Sales were highest on Friday and Monday, while Wednesday recorded the lowest sales.

## Business Insight

Customer purchasing behavior varies throughout the week, indicating that certain days generate stronger demand than others.

## Business Recommendation

Schedule promotions and marketing campaigns around high-performing days while using targeted offers to stimulate sales on lower-performing weekdays.
