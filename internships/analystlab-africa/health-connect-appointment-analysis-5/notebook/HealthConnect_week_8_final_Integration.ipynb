## Step 5: Final Analytical Findings

The Week 8 analytical findings are based on the results that were validated through the Week 6 statistical analysis and Week 7 testing and refinement activities.

### Finding 1: Overall No-Show Rate

The overall No-Show Rate in the HealthConnect dataset is **48.5%**.

This provides the baseline for interpreting the other appointment segments examined throughout the analysis.

The rate is specific to the synthetic HealthConnect dataset and should not be treated as a real-world healthcare benchmark.



### Finding 2: Longer Booking Lead Time Is Associated with Higher No-Show Rates

No-Show Rate increased across the booking lead-time groups:

| Booking Lead Time | No-Show Rate |
|---|---:|
| Same week (0–7 days) | 27.8% |
| 2–3 weeks (8–21 days) | 37.2% |
| 3+ weeks (22–60 days) | 56.9% |

The difference between the shortest and longest lead-time groups is **29.1 percentage points**.

Statistical validation found an association between booking lead time and appointment outcome:

- χ² = **280.09**
- p < **0.0001**
- Cramér's V = **0.167**

This indicates that booking lead time is an important observed factor in the dataset, although the analysis does not establish causation.



### Finding 3: Previous No-Show History Is Associated with Higher No-Show Rates

Appointments involving patients with at least one previous No-Show had a higher observed No-Show Rate than appointments with no previous No-Show history:

| Previous No-Show History | No-Show Rate |
|---|---:|
| No previous No-Show | 43.5% |
| Previous No-Show | 55.4% |

The difference is **11.9 percentage points**.

Statistical validation found an association between previous No-Show history and appointment outcome:

- χ² = **69.86**
- p < **0.0001**
- Cramér's V = **0.118**

This supports retaining previous No-Show history as an important analytical factor, while avoiding the interpretation that it independently causes future No-Shows.



### Finding 4: The Combined Lead Time × Previous No-Show Segmentation Provides Stronger Separation

Combining booking lead time with previous No-Show history produced the following observed rates:

| Lead Time | No Previous No-Show | Previous No-Show |
|---|---:|---:|
| Same week | 21.8% | 36.2% |
| 2–3 weeks | 34.1% | 41.9% |
| 3+ weeks | 51.7% | 64.1% |

The highest observed Data Analytics segment was **64.1%**:

**3+ weeks (≥22 days) + Previous No-Show**

The lowest observed segment was **21.8%**:

**Same week + No Previous No-Show**

This produces an observed difference of **42.3 percentage points**.

The combined six-category analysis had a Cramér's V of **0.189**, compared with **0.167** for booking lead time alone and **0.118** for previous No-Show history alone.

Therefore, among the tested segmentation approaches, the combined analysis provided the strongest observed association with appointment outcome.

This should be interpreted as an analytical segmentation result, not as proof of a causal interaction or guaranteed predictive performance.



### Finding 5: The Main Patterns Remained Visible Across Additional Segments

Week 6 and Week 7 testing examined whether the key patterns were concentrated in only one appointment type, distance group, reminder group, or gender group.

The longer-lead-time pattern remained visible across all four appointment types and across the reminder-status groups.

The previous-No-Show pattern also remained visible across the four appointment types.

However, some smaller subgroups showed less consistent patterns. For example, the **Prefer not to say** gender group showed a reversal in the 2–3 week band. Because this group contained only 108 records overall and was further divided across multiple combinations, this result should be interpreted cautiously.

The main findings should therefore be treated as validated patterns in the dataset rather than universal rules for every subgroup.



### Final Analytical Conclusion

The validated analysis indicates that **booking lead time and previous No-Show history are two of the most important observed factors examined in the HealthConnect dataset**.

Longer booking lead times were associated with higher No-Show Rates, and previous No-Show history was also associated with higher observed No-Show Rates. Combining these factors produced stronger separation between observed appointment segments than either factor considered alone.

These findings provide the analytical basis for the final business recommendations and can also support interpretation of the Data Science modelling component.

The findings remain observational and are based on a synthetic dataset. They should therefore be used as decision-support evidence within the HealthConnect project rather than as proof of causation or real-world patient behaviour.

## Step 6: Final Business Insights & Recommendations

### 6.1 Final Business Insights

#### Insight 1: Longer Booking Lead Times Represent an Important No-Show Segment

Appointments booked 3+ weeks in advance had an observed No-Show Rate of **56.9%**, compared with **27.8%** for appointments booked within the same week.

This suggests that appointments scheduled further in advance may require additional attention during appointment management.

**Business implication:** HealthConnect could consider reviewing how far in advance appointments are scheduled and whether additional confirmation or follow-up processes are appropriate for longer-lead-time appointments.



#### Insight 2: Previous No-Show History Identifies a Higher-Rate Segment

Patients with at least one previous No-Show had an observed No-Show Rate of **55.4%**, compared with **43.5%** among patients without a previous No-Show.

**Business implication:** Previous appointment history can be used as an analytical segmentation factor when considering where additional appointment-support attention may be useful.

This should be treated as a risk indicator within the analysis rather than as a reason to make assumptions about individual patients.



#### Insight 3: Combining Lead Time and Previous No-Show History Provides Greater Segment Separation

The combination of booking lead time and previous No-Show history produced a wider range of observed No-Show Rates than either factor considered separately.

The highest observed Data Analytics segment was:

**3+ weeks (≥22 days) + Previous No-Show — 64.1%**

This was compared with:

**Same week + No Previous No-Show — 21.8%**

**Business implication:** HealthConnect could use the combined segmentation as a decision-support framework for prioritising appointment-support activities, subject to further validation.



#### Insight 4: Reminder Results Should Be Interpreted Carefully

The observed No-Show Rate was:

- **51.4%** when no reminder was sent.
- **47.4%** when a reminder was sent.

Among appointments where reminders were sent, the observed rates were:

- SMS — **45.8%**
- Email — **48.4%**
- WhatsApp — **49.8%**

The differences are observational and do not establish that one reminder method causes better attendance.

**Business implication:** Reminder processes should continue to be monitored, but the current analysis does not provide sufficient evidence to recommend one channel as definitively more effective.



#### Insight 5: Distance Provides a Supporting Segmentation Factor

Observed No-Show Rates increased across the distance groups:

- Less than 5 km — **46.5%**
- 5–15 km — **47.3%**
- 15 km+ — **54.1%**

Distance therefore provides additional context for understanding appointment attendance patterns.

**Business implication:** Distance may be useful as a supporting factor when examining appointment-support needs, while recognising that the Week 6 modelling analysis did not show additional predictive value from distance in the reported model setup.



### 6.2 Final Business Recommendations

#### Recommendation 1: Prioritise Longer-Lead-Time Appointments for Additional Support

HealthConnect could consider introducing additional confirmation or appointment-support processes for appointments scheduled several weeks in advance.

The purpose would be to maintain patient engagement between booking and appointment date.

This recommendation is based on the observed increase in No-Show Rate across lead-time groups.



#### Recommendation 2: Consider Previous No-Show History in Appointment-Support Strategies

HealthConnect could consider previous No-Show history as one factor when deciding where additional appointment-support attention may be appropriate.

This should be applied as a support-oriented segmentation approach rather than as a judgement about individual patients.



#### Recommendation 3: Consider Combined Segmentation for Decision Support

The Lead Time × Previous No-Show segmentation can be used as a framework for identifying appointment groups with higher observed No-Show Rates.

The segmentation should remain clearly defined, including:

- Lead-time threshold
- Previous No-Show condition
- Appointment outcomes included in the denominator

This is particularly important because the Week 7 cross-track testing demonstrated that different segment definitions can produce different reported rates.



#### Recommendation 4: Continue Monitoring Reminder Performance

HealthConnect should continue monitoring reminder delivery and appointment outcomes by reminder status and channel.

The current analysis does not establish causality, so future testing should be used to determine whether changes to reminder processes are associated with measurable improvements in attendance.



#### Recommendation 5: Use the Dashboard as an Ongoing Decision-Support Tool

The Power BI dashboard can provide HealthConnect stakeholders with a way to monitor No-Show patterns across lead time, previous No-Show history, distance, reminder information, appointment type, and gender.

The dashboard should be used to support investigation and decision-making rather than as a standalone prediction or explanation of individual patient behaviour.

- Evidence on the relationship between booking lead time and appointment outcomes.

## Step 8: Analytical Limitations & Remaining Issues

### 8.1 Analytical Limitations

The following limitations remain relevant to the final HealthConnect Data Analytics component.

**1. Synthetic Dataset**

The HealthConnect dataset is synthetic. Therefore, the observed patterns should not automatically be treated as representative of real-world patient behaviour or actual clinic operations.

**2. Appointment History Limitations**

The `previous_appointments` and `previous_no_shows` fields may not represent fully realistic patient histories. Findings involving previous No-Show history should therefore be interpreted within the context of the available dataset.

**3. Missing Values**

Some variables contain missing values:

- `reminder_channel`: 1,366 records (27.3%). This missingness is structural because these records correspond to appointments where no reminder was sent.
- `distance_to_clinic_km`: 90 records (1.8%).
- `waiting_time_minutes`: 60 records (1.2%).

These missing values were documented during the earlier data-quality assessment.

**4. Association Does Not Establish Causation**

The analysis identifies observed relationships between variables and No-Show outcomes. It does not establish that longer booking lead time, previous No-Show history, reminders, distance, or other factors directly cause an appointment to be missed.

**5. Statistical Significance Does Not Equal Predictive Performance**

The statistical tests confirmed associations within the dataset, but statistical significance alone does not demonstrate predictive performance in real-world settings. The analytical findings should therefore continue to be considered alongside Data Science model validation.

**6. Segment Definitions Can Affect Reported Rates**

The Week 7 cross-track testing demonstrated that different lead-time thresholds and denominator definitions can produce different No-Show Rates.

For this reason, segment definitions should always be stated alongside reported percentages.

**7. Small Subgroup Sizes**

Some subgroup analyses contain relatively small numbers of observations. For example, the **Prefer not to say** gender group contains 108 records, with further splits across lead-time and previous No-Show combinations.

These smaller groups should therefore be interpreted cautiously.

**8. Reminder Analysis Remains Observational**

Differences in No-Show Rates across reminder status and reminder channels were observed, but the analysis does not establish that sending a reminder or using a particular reminder channel caused the observed differences.



### 8.2 Remaining Issues

No confirmed KPI calculation errors remain after the Week 7 testing and re-testing activities.

However, the following issues should remain documented for future work:

- Continue using explicit segment definitions and denominator rules when comparing results across tracks.
- Continue validating the predictive usefulness of the analytical findings with Data Science.
- Avoid treating reminder comparisons as evidence of reminder effectiveness without further testing.
- Interpret small subgroup patterns cautiously and avoid drawing broad conclusions from them.
- Reassess the findings using representative real-world clinic data if such data becomes available.
- Continue validating the dashboard when new data, configurations, or filter combinations are introduced.



### 8.3 Overall Status

The remaining limitations and issues do **not prevent final integration** of the Data Analytics component.

The dashboard calculations and core analytical findings have been validated, the identified dashboard issue was resolved and re-tested, and the cross-track segment discrepancy was clarified through independent reproduction.

The remaining considerations mainly affect **how the findings should be interpreted and applied**, rather than indicating unresolved calculation errors.

The Data Analytics component is therefore ready to proceed to final integration, presentation, and project showcase.

# 10. Final End-to-End Integration Walkthrough

## 10.1 The Problem

HealthConnect is addressing the problem of **missed appointments and the need to improve the patient support experience**.

The Data Analytics contribution focuses on understanding appointment attendance patterns and identifying factors associated with higher observed No-Show Rates.

## 10.2 The Data

The Data Analytics track used the **HealthConnect appointment dataset**, containing 5,000 appointment records and 18 variables.

The analysis examined factors including:

- Booking lead time
- Previous No-Show history
- Distance to the clinic
- Reminder status and channel
- Appointment type
- Gender
- Appointment outcomes

The **HealthConnect Data Dictionary** was also used to understand the meaning and structure of the variables.

## 10.3 The Analytics

The Data Analytics analysis identified several important observed patterns.

The overall No-Show Rate was **48.5%**.

Longer booking lead times were associated with higher observed No-Show Rates:

- Same week: **27.8%**
- 2–3 weeks: **37.2%**
- 3+ weeks: **56.9%**

Previous No-Show history also showed an association with higher No-Show Rates:

- No previous No-Show: **43.5%**
- Previous No-Show: **55.4%**

Combining booking lead time and previous No-Show history produced clearer separation between appointment segments.

The highest observed Data Analytics segment was **64.1%**, defined as:

- **Lead time:** 3+ weeks (≥22 days)
- **Previous No-Show:** At least 1
- **Denominator:** All appointment outcomes included

These findings were statistically validated, and the dashboard calculations and visuals were tested against the underlying dataset.

## 10.4 The Predictive Component

The Data Science track contributed predictive modelling using the HealthConnect appointment data.

The modelling work included **Logistic Regression** and **Random Forest** approaches and evaluated model performance using metrics such as:

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC

The Data Analytics findings, particularly **booking lead time and previous No-Show history**, were also relevant to the Data Science modelling and interpretation.

Cross-track testing was performed to ensure that important analytical segment definitions were clearly understood and reproducible.

## 10.10 Limitations

The final solution has several limitations:

- The HealthConnect dataset is **synthetic**, so the findings should not automatically be treated as representative of real-world patient behaviour.
- The analysis identifies **associations rather than causation**.
- Some subgroup analyses contain relatively small groups and should therefore be interpreted cautiously.
- Some variables contain missing values, including `reminder_channel`, `distance_to_clinic_km`, and `waiting_time_minutes`.
- Different segment definitions can produce different reported rates, so thresholds and denominator rules must be clearly documented.
- Data Analytics findings do not by themselves establish predictive performance.
- The predictive component has its own modelling limitations and should be interpreted using its reported validation results.
- Dashboard validation covers the configurations tested during Week 7 and Week 8 and does not guarantee behaviour under every possible future data or filter state.
