## Step 3: KPI Validation

The five KPI cards displayed on the Week 6 Power BI dashboard were independently recalculated using the raw HealthConnect appointment dataset.

The recalculations were performed in this Week 7 testing notebook using Python and pandas. The calculations followed the same underlying definitions used for the Power BI measures, allowing the dashboard values to be compared directly with independently calculated results.

The validation focused on confirming that the correct numerator, denominator, and segment definitions were being used for each KPI.

```python
import pandas as pd
df = pd.read_csv("HealthConnect_Appointment_Data.csv")

# Independent KPI validation

# 1. Overall No-Show Rate
overall_no_show_rate = (
    df["appointment_outcome"].eq("No-Show").mean()
)

# 2. No-Show Rate for appointments with no reminder
no_reminder = df[df["reminder_sent"] == "No"]

no_reminder_rate = (
    no_reminder["appointment_outcome"].eq("No-Show").mean()
)

# 3. No-Show Rate for appointments with previous No-Shows
previous_no_show = df[df["previous_no_shows"] > 0]

previous_no_show_rate = (
    previous_no_show["appointment_outcome"].eq("No-Show").mean()
)

# 4. No-Show Rate for 3+ week lead time
three_plus_weeks = df[df["booking_lead_days"] >= 22]

three_plus_week_rate = (
    three_plus_weeks["appointment_outcome"].eq("No-Show").mean()
)

# 5. No-Show Rate for 3+ weeks + Previous No-Show
highest_observed_segment = df[
    (df["booking_lead_days"] >= 22) &
    (df["previous_no_shows"] > 0)
]

highest_segment_rate = (
    highest_observed_segment["appointment_outcome"].eq("No-Show").mean()
)

# Display results
print(f"Overall No-Show Rate: {overall_no_show_rate:.2%}")
print(f"No Reminder: {no_reminder_rate:.2%}")
print(f"Previous No-Show: {previous_no_show_rate:.2%}")
print(f"3+ Week Lead Time: {three_plus_week_rate:.2%}")
print(f"Highest Observed Segment: {highest_segment_rate:.2%}")

```

### Observation

The independently calculated KPI values matched the Power BI dashboard values after rounding to one decimal place.

The validation confirmed that:

- The Overall No-Show Rate is **48.46%**, displayed as **48.5%** in Power BI.
- The No Reminder rate is **51.39%**, displayed as **51.4%**.
- The Previous No-Show rate is **55.41%**, displayed as **55.4%**.
- The 3+ Week Lead Time rate is **56.94%**, displayed as **56.9%**.
- The Highest Observed Segment rate is **64.09%**, displayed as **64.1%**.

No discrepancy was identified in the five KPI calculations.

The same denominator definition used in the Data Analytics analysis was applied, meaning cancelled appointments remain included in the denominator.

## Step 4: Analytical Validation

The key analytical findings from Weeks 5 and 6 were independently validated using the raw HealthConnect appointment dataset.

The purpose of this testing was to confirm that the main No-Show patterns identified during the earlier analysis remained consistent when recalculated independently.

The validation focused on:

- Booking Lead Time and No-Show Rate
- Previous No-Show History and No-Show Rate
- Combined Lead Time × Previous No-Show analysis
- Chi-square tests and Cramér's V used to validate the main associations

The results were compared with the Week 5 and Week 6 findings to identify any inconsistencies or calculation errors.

```python
# Create the same lead-time bands used in the Week 5/6 analysis

df["lead_time_band"] = pd.cut(
    df["booking_lead_days"],
    bins=[-1, 7, 21, 60],
    labels=[
        "Same week (0-7 days)",
        "2-3 weeks (8-21 days)",
        "3+ weeks (22-60 days)"
    ]
)

# Calculate No-Show Rate by lead-time band

lead_time_validation = (
    df.groupby("lead_time_band", observed=False)["appointment_outcome"]
      .apply(lambda x: (x == "No-Show").mean() * 100)
      .round(2)
)

lead_time_validation
```

### Observation

The independently recalculated No-Show rates were **27.81%**, **37.21%**, and **56.94%** for Same Week, 2–3 Weeks, and 3+ Weeks respectively.

These results match the Week 5/6 findings after rounding.

The same pattern remains visible: the observed No-Show rate increases as booking lead time increases.

**Validation result: PASS**

```python
# Create previous No-Show groups
import numpy as np

df["previous_no_show_group"] = np.where(
    df["previous_no_shows"] > 0,
    "Previous No-Show",
    "No Previous No-Show"
)

# Calculate No-Show Rate by previous No-Show history

previous_no_show_validation = (
    df.groupby("previous_no_show_group")["appointment_outcome"]
      .apply(lambda x: (x == "No-Show").mean() * 100)
      .round(2)
)

previous_no_show_validation
```

### Observation

The independently recalculated No-Show rates were **43.51%** for appointments with no previous No-Show history and **55.41%** for appointments with previous No-Show history.

These results match the Week 5/6 findings after rounding.

The higher observed No-Show rate among appointments with previous No-Show history therefore remains present in the raw dataset.

**Validation result: PASS**

```python
# Combine lead time and previous No-Show history

df["lead_time_previous_segment"] = (
    df["lead_time_band"].astype(str)
    + " + "
    + df["previous_no_show_group"]
)

# Calculate No-Show Rate for each combined segment

combined_validation = (
    df.groupby("lead_time_previous_segment", observed=False)["appointment_outcome"]
      .apply(lambda x: (x == "No-Show").mean() * 100)
      .round(2)
      .sort_values()
)

combined_validation
```

### Observation

The combined Lead Time × Previous No-Show analysis reproduced the Week 5/6 pattern.

The lowest observed No-Show rate was **21.77%** for appointments with Same Week lead time and no previous No-Show history.

The highest observed No-Show rate was **64.09%** for appointments with **3+ weeks lead time and previous No-Show history**.

The dashboard displays this highest observed segment as **64.1%**, which is consistent with the independent calculation after rounding.

**Validation result: PASS**

```python
from scipy.stats import chi2_contingency
import numpy as np

# Booking Lead Time vs Appointment Outcome

lead_time_table = pd.crosstab(
    df["lead_time_band"],
    df["appointment_outcome"]
)

chi2_lead, p_lead, dof_lead, expected_lead = chi2_contingency(
    lead_time_table
)

n_lead = lead_time_table.values.sum()
rows_lead, cols_lead = lead_time_table.shape

cramers_v_lead = np.sqrt(
    chi2_lead / (n_lead * min(rows_lead - 1, cols_lead - 1))
)

print("Booking Lead Time")
print("Chi-square:", round(chi2_lead, 2))
print("p-value:", p_lead)
print("Degrees of freedom:", dof_lead)
print("Cramér's V:", round(cramers_v_lead, 3))
```

```python
# Previous No-Show History vs Appointment Outcome

previous_table = pd.crosstab(
    df["previous_no_show_group"],
    df["appointment_outcome"]
)

chi2_previous, p_previous, dof_previous, expected_previous = chi2_contingency(
    previous_table
)

n_previous = previous_table.values.sum()
rows_previous, cols_previous = previous_table.shape

cramers_v_previous = np.sqrt(
    chi2_previous / (n_previous * min(rows_previous - 1, cols_previous - 1))
)

print("Previous No-Show History")
print("Chi-square:", round(chi2_previous, 2))
print("p-value:", p_previous)
print("Degrees of freedom:", dof_previous)
print("Cramér's V:", round(cramers_v_previous, 3))
```

```python
# Combined Lead Time × Previous No-Show vs Appointment Outcome

combined_table = pd.crosstab(
    df["lead_time_previous_segment"],
    df["appointment_outcome"]
)

chi2_combined, p_combined, dof_combined, expected_combined = chi2_contingency(
    combined_table
)

n_combined = combined_table.values.sum()
rows_combined, cols_combined = combined_table.shape

cramers_v_combined = np.sqrt(
    chi2_combined / (n_combined * min(rows_combined - 1, cols_combined - 1))
)

print("Combined Lead Time × Previous No-Show")
print("Chi-square:", round(chi2_combined, 2))
print("p-value:", p_combined)
print("Degrees of freedom:", dof_combined)
print("Cramér's V:", round(cramers_v_combined, 3))
```

### Analytical Validation Conclusion

The Week 5 and Week 6 analytical findings were successfully reproduced using independent calculations from the raw dataset.

The validation confirmed that:

- Booking Lead Time remains associated with appointment outcome (χ² = **280.09**, Cramér's V = **0.167**).
- Previous No-Show History remains associated with appointment outcome (χ² = **69.86**, Cramér's V = **0.118**).
- The combined Lead Time × Previous No-Show analysis produced the highest Cramér's V among these tested analyses (χ² = **358.52**, Cramér's V = **0.189**).
- The highest observed segment remains **3+ weeks + Previous No-Show**, with a No-Show Rate of **64.09%**, displayed as **64.1%** in Power BI.

No analytical discrepancy was identified during this validation.

These results confirm that the main Week 6 findings are reproducible from the underlying dataset. The statistical results indicate association, not causation, and the Cramér's V values should not be interpreted as predictive-model performance.

**Degrees of freedom note:** The combined Lead Time × Previous No-Show test has **10 degrees of freedom** because the combined variable contains **6 categories** and appointment outcome contains **3 categories**. Therefore, `(6 − 1) × (3 − 1) = 10`. This differs from the Lead Time test (4 df) and Previous No-Show test (2 df) because those variables have fewer categories.

## Step 5: Segment Testing

The main Week 6 findings were further tested across relevant appointment and patient segments to determine whether the observed patterns remained consistent.

The testing focused on the relationship between Booking Lead Time and No-Show Rate across:

- Distance to Clinic
- Reminder Status

The combined Lead Time × Previous No-Show finding was also reviewed across Appointment Type.

The purpose of this testing was not to claim that the relationships are independent of other factors, but to determine whether the main observed patterns remained visible across different segments.

```python
# No-Show Rate by Appointment Type and Lead Time

lead_time_by_appointment = pd.crosstab(
    df["appointment_type"],
    df["lead_time_band"],
    values=df["appointment_outcome"].eq("No-Show"),
    aggfunc="mean"
) * 100

lead_time_by_appointment.round(2)
```

### Observation: Appointment Type

The No-Show Rate increased as Booking Lead Time increased across all four appointment types.

The No-Show Rate increased from the Same Week category to the 3+ Week category for:

- Diagnostic Test
- Follow-up
- General Consultation
- Specialist Consultation

The increase ranged from approximately **26.1 to 30.5 percentage points** across the four appointment types.

This confirms that the main Lead Time pattern identified in Week 6 remains visible across appointment types.

**Validation result: PASS**

```python
# No-Show Rate by Distance Band and Lead Time
df["distance_band"] = pd.cut(
    df["distance_to_clinic_km"],
    bins=[0, 5, 15, 45],
    labels=["<5km", "5-15km", "15km+"]
)

lead_time_by_distance = pd.crosstab(
    df["distance_band"],
    df["lead_time_band"],
    values=df["appointment_outcome"].eq("No-Show"),
    aggfunc="mean"
) * 100

lead_time_by_distance.round(2)
```

### Observation: Distance to Clinic

The No-Show Rate increased with longer Booking Lead Time across all three distance groups.

The 3+ Week category had the highest observed No-Show Rate within each distance group.

The increase from Same Week to 3+ Weeks ranged from approximately **25.9 to 33.7 percentage points**.

The results therefore support the Week 6 observation that the Lead Time pattern remains visible across different distance groups.

**Validation result: PASS**

```python
# No-Show Rate by Reminder Status and Lead Time

lead_time_by_reminder = pd.crosstab(
    df["reminder_sent"],
    df["lead_time_band"],
    values=df["appointment_outcome"].eq("No-Show"),
    aggfunc="mean"
) * 100

lead_time_by_reminder.round(2)
```

### Observation: Reminder Status

The No-Show Rate increased with longer Booking Lead Time in both reminder-status groups.

For appointments where no reminder was sent, the rate increased from **29.9%** in the Same Week group to **59.6%** in the 3+ Week group.

For appointments where a reminder was sent, the rate increased from **27.1%** to **55.9%**.

The Lead Time pattern therefore remained visible in both reminder-status groups.

This does not establish that reminder status has no effect or that the two factors are independent. It only confirms that the observed Lead Time pattern remains present within both reminder-status groups.

**Validation result: PASS**

```python
# No-Show Rate by Appointment Type,
# Lead Time and Previous No-Show History

combined_by_appointment = pd.crosstab(
    df["appointment_type"],
    [df["lead_time_band"], df["previous_no_show_group"]],
    values=df["appointment_outcome"].eq("No-Show"),
    aggfunc="mean"
) * 100

combined_by_appointment.round(2)
```

The **3+ weeks + Previous No-Show** segment had the highest observed No-Show Rate within each appointment type:

The rates ranged from **61.9% to 68.0%**, showing that the high observed No-Show pattern was present across all four appointment types.

One smaller subgroup showed a reversal in the 2–3 week category for Diagnostic Tests. This subgroup should be interpreted cautiously because of its smaller sample size.

Overall, the highest-risk observed segment identified in the broader analysis remained visible across appointment types.

**Validation result: PASS**

### Segment Testing Conclusion

The Week 6 analytical findings were tested across Appointment Type, Distance to Clinic, and Reminder Status.

The testing confirmed that:

- The Lead Time pattern remained visible across distance groups.
- The Lead Time pattern remained visible across both reminder-status groups.

## Step 15: Limitations

Although the Week 7 testing confirmed the accuracy of the main dashboard calculations and analytical findings, several limitations remain relevant when interpreting the HealthConnect analysis.

### 1. Synthetic Dataset

The HealthConnect dataset is a synthetic dataset. Therefore, the observed No-Show patterns should not automatically be treated as representative of real-world patient behaviour or actual HealthConnect operations.

The overall No-Show Rate of **48.5%**, for example, should be interpreted as a characteristic of this dataset rather than as a real-world benchmark.



### 2. Previous Appointment History May Not Represent Real Patient Histories

The `previous_appointments` and `previous_no_shows` fields may not represent fully coherent longitudinal patient histories.

Although the Week 4 validation confirmed that `previous_no_shows` does not exceed `previous_appointments`, the dataset structure does not allow the analysis to independently verify the complete sequence of each patient's historical appointments.

Therefore, findings involving previous No-Show history should be interpreted within the limits of the available dataset.



### 3. Missing Values Remain in Some Variables

Some variables contain missing values:

- `reminder_channel`: **1,366 records (27.3%)**
- `distance_to_clinic_km`: **90 records (1.8%)**
- `waiting_time_minutes`: **60 records (1.2%)**

The missingness in `reminder_channel` is structurally related to appointments where no reminder was sent.

The relatively small amount of missingness in distance and waiting time should still be considered when interpreting analyses involving those variables.



### 4. Observational Analysis Does Not Establish Causation

The analysis identifies associations and observed differences between groups. It does not establish that a particular factor directly causes a No-Show.

For example, the lower observed No-Show Rate among appointments where reminders were sent does not by itself prove that reminders caused the difference.

Further controlled analysis or operational testing would be required to establish causal effects.



### 5. Statistical Association Does Not Equal Predictive Performance

The Week 6 statistical tests confirmed associations between appointment outcome and factors such as booking lead time and previous No-Show history.

However, statistical association does not by itself demonstrate that a variable will produce strong predictions for individual appointments.

The Data Science modelling results provide additional predictive context, but model performance remains dependent on the available features, dataset, modelling approach, and validation methodology.



### 6. Segment Definitions Affect Reported Rates

Week 7 cross-track testing demonstrated that No-Show Rates can change when the segment definition changes.

The original **64.1%** Data Analytics result used a lead-time threshold of `booking_lead_days ≥ 22` and included Cancelled appointments in the denominator.

The Data Science analysis used `booking_lead_days > 30` and originally excluded Cancelled appointments.

The testing reproduced the Data Science results under its stated definitions (**70.52% excluding Cancelled and 67.88% including Cancelled**).

Therefore, No-Show Rates from different analyses should only be compared when the segment criteria and denominator definitions are clearly aligned.



### 7. Small Subgroups Require Cautious Interpretation

Some categories contain substantially fewer observations than others.

For example, the **Prefer not to say** gender group contains only **108 records**, which are further divided across multiple lead-time and previous No-Show combinations.

The reversed pattern observed in the 2–3 week segment should therefore be interpreted cautiously and should not be treated as a general behavioural pattern without additional data.



### 8. Dashboard Validation Covers the Tested Configuration

The Week 7 testing confirmed the five KPI cards, six comparison visuals, slicers, combined analysis, and Lead Time Band ordering under the tested conditions.

However, dashboard accuracy can change if the underlying dataset, DAX measures, relationships, filters, or segment definitions are modified in the future.

Any substantial changes to the dashboard should therefore be followed by another validation cycle.



### Overall Limitation

The Week 7 testing increased confidence in the accuracy and consistency of the current analysis, but the findings remain limited by the synthetic dataset, available variables, missing values, observational design, subgroup sizes, and differences in analytical definitions.

### 1. Build on the Validated Lead Time and Previous No-Show Findings

Week 7 confirmed the consistency of the observed relationship between booking lead time, previous No-Show history, and appointment outcomes.

**Recommendation:**
- Carry the validated Lead Time × Previous No-Show segmentation forward into Week 8.
- Maintain the documented segment definitions when using or comparing these results.
- Continue evaluating whether these factors remain useful as the HealthConnect solution develops.



### 2. Maintain Clear Analytical Definitions Across Tracks

The Week 7 cross-track testing showed that differences in segment thresholds and denominator definitions can produce different No-Show Rates from the same underlying dataset.

**Recommendation:**
- Define analytical criteria before cross-track comparisons are performed.
- Document lead-time thresholds, No-Show conditions, and denominator rules alongside reported metrics.
- Use consistent definitions when different tracks need to compare or integrate analytical outputs.



### 3. Continue Cross-Track Integration and Validation

The Data Analytics and Data Science collaboration helped validate the reported segment calculations and identify differences in analytical definitions.

**Recommendation:**
- Continue sharing validated findings and clearly defined analytical outputs with relevant HealthConnect tracks.
- Where another track uses an analytical result from Data Analytics, provide the calculation definition and denominator alongside the result.
- Use cross-track testing when an output becomes a dependency for another part of the HealthConnect solution.



### 4. Further Evaluate Predictive Use of the Validated Segments

The combined Lead Time × Previous No-Show analysis showed the strongest observed association among the tested segmentation approaches, while Data Science modelling provided additional predictive context.

**Recommendation:**
- Continue evaluating the combined segment in future predictive analysis.
- Compare its usefulness with individual features such as booking lead time and previous No-Show history.
- Avoid treating the segment as a confirmed predictive or causal factor without further model validation.



### 5. Evaluate Reminder Effectiveness With Future Operational Data

The current analysis identified differences between reminder groups and channels but did not establish causal effectiveness.

**Recommendation:**
- If operational data becomes available, evaluate reminder effectiveness using a suitable testing or evaluation approach.
- Continue monitoring reminder status and channel alongside appointment outcomes.
- Avoid making operational decisions about the most effective reminder channel based solely on the current dataset.



### 6. Reassess Findings Using More Representative Data

The current HealthConnect dataset is synthetic, which limits how confidently the observed patterns can be applied to real-world operations.

**Recommendation:**
- If real operational data becomes available, reassess the validated Week 7 findings.
- Compare whether the relationships involving lead time, previous No-Show history, distance, and reminders remain consistent.
- Recalculate KPIs using the same documented definitions to maintain comparability.



### 7. Preserve the Validated Dashboard as a Baseline

The Week 7 testing confirmed the main KPI cards, comparison visuals, slicers, calculations, and Lead Time Band ordering.

**Recommendation:**
- Treat the validated Week 7 dashboard as a baseline for future development.
- Re-test KPI calculations and filters whenever major changes are made.
- Preserve the current KPI and segment definitions unless a documented analytical change is introduced.



### Overall Week 8 Direction

Week 8 should build on the validated Week 7 outputs rather than restarting the analysis.

The main priorities should be to:

1. **Carry forward the validated analytical findings.**
2. **Maintain consistent definitions across HealthConnect tracks.**
3. **Continue cross-track integration where analytical outputs become dependencies.**
4. **Further evaluate the predictive usefulness of the combined segmentation.**
5. **Reassess the findings when more representative data becomes available.**
6. **Use the validated Power BI dashboard as the baseline for future improvements.**

These recommendations provide a transition from Week 7 testing and validation toward continued HealthConnect integration and development in Week 8.
